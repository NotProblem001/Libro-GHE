## ❓ Si algo sale mal

| Mensaje | Qué hacer |
|---|---|
| ❌ *Corrija lo siguiente en data/PLANTILLA_MES.xlsx…* | Abra la plantilla, corrija la hoja y fila indicadas, guarde, cierre y presione ▶ otra vez. |
| ❌ *…está abierto en Excel* | Cierre ese archivo en Excel y presione ▶ otra vez. |
| ❌ *No se encontró ningún horario anterior* | Copie el Excel del mes pasado en `data/` con el nombre `horario_mes_anterior.xlsx`. |
| ⚠️ *Hay N alertas para revisar* | El horario **sí se creó**. Revise la hoja **ALERTAS** del Excel: indica días y horas donde no se cumple alguna regla. |
| `ModuleNotFoundError: motor_turnos` | El archivo `motor_turnos.py` debe estar en la misma carpeta que este libro. |
| La plantilla quedó dañada | Bórrela y presione ▶: se crea una nueva con los valores originales (tendrá que volver a llenar EJECUTIVOS). |
